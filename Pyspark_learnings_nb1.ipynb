```python
from pyspark.sql import SparkSession, Row
from pyspark.sql.functions import col, sum, when, avg, max, min, count, row_number, rank, dense_rank, lag
from pyspark.sql.window import Window
from pyspark.sql.functions import to_date, to_timestamp, year, month, dayofmonth,hour, minute, datediff, \
date_add, date_sub, current_date, date_diff, date_format
spark = SparkSession.builder.appName("learningsession").getOrCreate()
```

```python
#start with D12
data = [("a", "el", 11),
        ("b", "el", 12),
        ("c", "fur", 13),
        ("d", "fur", 14),
        ("e", "el", 15),
        ("f", "fur", 16),
        ("e","fur", 22),
        ("f", "el", 24)
]
df  = spark.createDataFrame(data, ["Customer", "category", "Amount"])
df.show()
```

```python
df.groupBy("category").sum("amount").show()
```

```python
df.groupBy("customer").sum("amount").show()
```

```python
#cntd D12S3
df.groupBy("Customer").agg(sum("Amount").alias("Totalspent"),\
        avg("Amount").alias("Avgspent"),\
        max("Amount").alias("Maxspent"),\
        min("Amount").alias("Minspent")).show()
```

```python
df.groupBy("Customer", "category").sum("Amount").show()
```

```python
cat_agg = df.groupBy("category").agg(\
             sum("Amount").alias("Totalspent"),\
             avg("Amount").alias("Avgspent"),\
             count("*").alias("Totalcount")\
             )
cat_agg.show()
```

```python
 cat_cus_df = df.groupBy("Customer", "category")\
               .agg(sum("Amount").alias("Totalspent"),\
                    max("Amount").alias("Maxamount"))
cat_cus_df.show()
```

```python
# D13 start with widow functs, window_spec
window_spec = Window.partitionBy("Customer").orderBy(df['Amount'])
type(window_spec)
```

```python
# ranking the DF.
df_ranks = df.withColumn("row_number", row_number().over(window_spec))\
             .withColumn("rank", rank().over(window_spec))\
             .withColumn("dense_rank", dense_rank().over(window_spec))
df_ranks.show()
```

```python
df_lag = df.withColumn("lag_amount", lag("Amount").over(window_spec))
df_lag.show()
```

```python
#D13 mini challenge
df_rank = df.withColumn("rank",rank().over(window_spec))
df_filtered_toppurchase = df_rank.filter(col("rank")==1).drop("rank")
df_filtered_toppurchase.show()
# check answer in cgpt
```

```python
#D13 mini challenge spend change
df_prev = df.withColumn("prev_col_amt", lag("Amount", 1).over(window_spec))
df_change= df_prev.withColumn("spendchange", col("Amount")- col("prev_col_amt"))
df_change.show()
```

```python
# D14 Timestamps and handling dates
data = [
    Row(Event="Order1", DateStr="2024-01-15", TimeStr="2024-01-15 10:30:00"),
    Row(Event="Order2", DateStr="2024-02-05", TimeStr="2024-02-05 13:45:30"),
    Row(Event="Order3", DateStr="2024-03-22", TimeStr="2024-03-22 18:00:00"),
]

df = spark.createDataFrame(data)
df_parsed = df.withColumn("Date", to_date("DateStr"))\
               .withColumn("Timestamp",to_timestamp("Timestr"))

df_parsed.show(truncate =False)

```

```python
#D14 s2 Extract date parts
df_extracted = df_parsed.select(
    "Event", "Date", "Timestamp",
    year("Date").alias("Year"),
    month("Date").alias("Month"),
    dayofmonth("Date").alias("Day"),
    hour("Timestamp").alias("Hour"),
    minute("Timestamp").alias("Minute")

)
df_extracted.show()
```

```python
#D14 S3 date arthimatics
df_dates = df_parsed.withColumn("Date_diff", date_diff(current_date(), "Date"))\
                  .withColumn("DeliverydateETA", date_add("Date", 7))
```

```python

# day14 mini challenge
df_month = df_parsed.withColumn("Monthname", date_format("Date", "MMMM"))
df_month.show()
# 2
df_weekend = df_parsed.withColumn("Weekendcol", date_format("Date" \
            ,"EEEE")).filter(col("Weekendcol").isin("Saturday","Sunday"))
df_weekend.show()
```

```python
from pyspark.sql.functions import col, date_format

df_weekend = (
    df_parsed
    .withColumn("Weekendcol", date_format("Date", "EEEE"))
    .filter(col("Weekendcol").isin("Saturday", "Sunday"))
)

df_weekend.show()
# check this
```

```python
# start d15
data_users = [
    ("u1", "Alice"),
    ("u2", "Bob"),
    ("u3", "Charlie")
]
df_users = spark.createDataFrame(data_users, ["UserID", "Name"])

data_orders = [
    ("u1", "Order1"),
    ("u2", "Order2"),
    ("u4", "Order3")   # note: u4 doesn't exist in users
]
df_orders = spark.createDataFrame(data_orders, ["UserID", "OrderID"])

df_users.show()
df_orders.show()
# cntd
```

```python
#joining
df_joined = df_users.join(df_orders,
                on=df_users.UserID == df_orders.UserID, # or simply on = "UserID"
                how= "inner"
                )
df_joined.show()
#
```

```python
# left join
df_left_joined = df_users.join(df_orders, on="UserId", how="left")
df_left_joined.show()
```

```python
# right join
df_righ_join = df_users.join(df_orders, on="UserId", how="right")
df_righ_join.show()
#now check '
```

```python
# mini challenge d15
df_semi_right = df_users.join(df_orders, on="UserID", how="left_semi")
df_semi_right.show()
# verify the answer with cgpt
```

```python
data_sales = [
    ("u1", "Electronics", 1200),
    ("u2", "Furniture", 800),
    ("u3", "Electronics", 600),
    ("u1", "Furniture", 300)
]
df_sales = spark.createDataFrame(data_sales, ["UserID", "Category", "Amount"])

data_discounts = [
    ("u1", "Electronics", 0.10),
    ("u1", "Furniture", 0.05),
    ("u2", "Furniture", 0.15),
    ("u4", "Electronics", 0.20)   # note: u4 not in sales
]
df_discounts = spark.createDataFrame(data_discounts, ["UserID", "Category", "Discount"])

df_sales.show()
df_discounts.show()
```

```python
# join on multiple coloumns
df_joined = df_sales.join(df_discounts, on=["UserID", "Category"])
df_joined.show()
```

```python
# day 16 mini challenge
df_joined = df_sales.join(df_discounts, on=["userID", "Category"])\
                     .withColumn("Discountedcol", col("Amount")* (1-col("Discount")))
df_joined.show()
```
